```python
import pandas as pd
from scipy import stats


cat_cols = df.select_dtypes(include=['object', 'category']).columns


results = []
for col in cat_cols:
    
    temp_df = df[[col, 'Ticket_Price_Cr']].dropna()
    groups = [temp_df[temp_df[col] == c]['Ticket_Price_Cr'] for c in temp_df[col].unique()]
    
    if len(groups) > 1:
        f_stat, p_val = stats.f_oneway(*groups)
        results.append({'Column': col, 'P-value': p_val, 'F-Statistic': f_stat})

anova_df = pd.DataFrame(results).sort_values('P-value')
print(anova_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create the plot
for cat in cat_list:
    plt.figure(figsize=(10, 6))
    plot = sns.barplot(x=df[cat], y=df['Ticket_Price_Cr'], data=anova_df, palette='viridis')

    # Add a significance threshold line at p = 0.05
    plt.axvline(0.05, color='red', linestyle='--', label='Significance (0.05)')

    # Labels and Title
    plt.title(f'Relation between {cat} vs Ticket_Price_Cr', fontsize=15)
    plt.xlabel(f'{cat}')
    plt.ylabel('Ticket_Price_Cr')
    plt.xticks(rotation=45)
    plt.legend()

    plt.show()
```
